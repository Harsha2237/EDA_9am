**Step-1: Import packages**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Step-2: Create a Df using list**

```python
names=['Ramesh','Suresh','Sathish']
#pd.DataFrame(data,index,columns)
pd.DataFrame(names)
```

**Step-3: I want to change the index**

```python
names=['Ramesh','Suresh','Sathish']
ids=['A','B','C']
pd.DataFrame(names,ids)
# pd.DataFrame(data,index)
```

**Step-4: add the columns**

```python
names=['Ramesh','Suresh','Sathish']
ids=['A','B','C']
cols=['Names']
pd.DataFrame(names,ids,cols)
```

**Step-5: Data and columns**

```python
names=['Ramesh','Suresh','Sathish']
cols=['Names']
pd.DataFrame(names,columns=cols)
```

**Step-6: Multiple columns**

```python
names=['Ramesh','Suresh','Sathish']
age=[20,22,24]
cols=['Names','Age']
df=pd.DataFrame(zip(names,age),
             columns=cols)
# pd.DataFrame(data,index)
df
```

**Step-7: Add a new column to alreday existed df**

- in above first we have taken data

- then created a dataframe

- now we want to add new data based on requirement

```python
city=['Hyd','Pune','Blr']
df['City']=city
df
```

```python
df['Marks']=[20,30,40]
```

```python
df
```

**Step-8: Update the column with new values**

```python
df['Marks']=[200,300,400]
```

```python
df
```

**Step-9: create empty dataframe and add the columns**

```python
df=pd.DataFrame()
df['Names']=['Ramesh','Suresh','Sathish']
df['Age']=[20,22,24]
df['City']=['Hyd','Pune','Blr']
df
```

**Step-10: Create a df using dictionary**

```python
d={'Names':['Ramesh','Suresh','Sathish'],
   'Age':[20,22,24],
   'City':['Hyd','Blr','Pune']}
df=pd.DataFrame(d)
df
```

```python
d={'Name':'Ramesh',
  'Age':20}
pd.DataFrame(d)
```

```python
d={'Name':'Ramesh',
  'Age':20}
pd.DataFrame(d,index=[0])
```
